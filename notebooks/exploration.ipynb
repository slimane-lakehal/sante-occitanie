```python
import pandas as pd
from pathlib import Path
import sys

# Path.cwd() returns the 'notebooks' directory, so .parent gets you 'sante-occitanie'
project_root = str(Path.cwd().parent)

# Append the project root if it is not already in sys.path
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from scripts.config import OCCITANIE_DEPTS
```

```python
RAW_DATA_DIR = Path('__file__').resolve().parent.parent / "data" / "raw"
raw_hospital = RAW_DATA_DIR / "hospital_covid_raw.csv"
```

```python
hospital_df = pd.read_csv(raw_hospital, sep=';')
hospital_df.head()
```

```python
df = hospital_df[hospital_df['sexe'] == 0]
df
```

```python
df = df[df['dep'].astype(str).str.zfill(2).isin(OCCITANIE_DEPTS)].copy()
df
```

```python
df["jour"] = pd.to_datetime(df["jour"])
```
